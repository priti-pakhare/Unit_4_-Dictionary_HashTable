```python
# Dictionary Application using Hash Tables

size = 10
keys = [None] * size
values = [None] * size


def hash_function(word):
    return sum(ord(ch) for ch in word) % size


def insert(word, meaning):
    index = hash_function(word)

    while keys[index] is not None and keys[index] != word:
        index = (index + 1) % size

    keys[index] = word
    values[index] = meaning
    print("Word added successfully!")


def search(word):
    index = hash_function(word)
    start = index

    while keys[index] is not None:
        if keys[index] == word:
            print("Meaning:", values[index])
            return

        index = (index + 1) % size

        if index == start:
            break

    print("Word not found!")


while True:
    print("\n--- DICTIONARY APPLICATION ---")
    print("1. Add Word")
    print("2. Search Word")
    print("3. Display Dictionary")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        word = input("Enter word: ").lower()
        meaning = input("Enter meaning: ")
        insert(word, meaning)

    elif choice == "2":
        word = input("Enter word to search: ").lower()
        search(word)

    elif choice == "3":
        print("\nWord\t\tMeaning")
        for i in range(size):
            if keys[i] is not None:
                print(keys[i], "\t\t", values[i])

    elif choice == "4":
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
